- *¿Qué establece el Principio de Invarianza respecto al tiempo de ejecución de dos implementaciones distintas para un mismo algoritmo?*  
**RESPUESTA:**  El Principio de Invarianza establece que el tiempo de ejecución de dos implementaciones distintas de un mismo algoritmo no diferirá más que en una constante multiplicativa. Formalmente, si se tiene un algoritmo y dos implementaciones suyas, $I_1$ e $I_2$ (que pueden ser en máquinas distintas o con códigos distintos), cuyos tiempos de ejecución son $T_1(n)$ y $T_2(n)$ respectivamente, el principio afirma que existe una constante real $c > 0$ y un número natural $n_0$ tales que para todo $n \ge n_0$ se cumple que:$$T_1(n) \le c \cdot T_2(n)$$

**RESPUESTA:** Piso ($\lfloor x \rfloor$) y Techo ($\lceil x \rceil$): Se usan para trabajar con enteros discretos: Piso define la operación módulo (residuos) y Techo calcula el número mínimo de bits necesarios ($\lceil \log_2 n \rceil$) y ajusta la división de tareas en algoritmos.Exponencial vs. Polinomial: Dado que $\lim_{n \to \infty} \frac{n^b}{a^n} = 0$ (para $a > 1$), toda exponencial termina superando a cualquier polinomio, lo que vuelve intratables a los algoritmos exponenciales para entradas grandes.

**RESPUESTA:** Conjunto finito, preciso, determinista y efectivo de instrucciones para transformar datos de entrada en una salida.
* **Problema del Paro:** Demostración de Turing de que es **indecidible** crear un programa general que determine si cualquier algoritmo se detendrá o entrará en un bucle infinito.

* **A Priori:** Evaluación teórica y matemática de la complejidad (independiente del hardware y lenguaje).

- *Explica en qué consisten el peor caso, mejor caso y caso promedio al momento de evaluar el comportamiento de un algoritmo ante distintas entradas.*  
**RESPUESTA:** * Peor caso (** **Worst case** **):** Máximo número de operaciones posibles. Representa la **cota superior** y da la garantía de que el algoritmo nunca tardará más que ese límite.
* **Mejor caso (** **Best case** **):** Mínimo número de operaciones posibles sobre la entrada ideal. Representa el **límite inferior** (suele ser el menos informativo en la práctica).
* **Caso promedio (** **Average case** **):** Comportamiento **esperado** sobre el conjunto de entradas posibles. Es la estimación más **realista**, aunque requiere análisis de probabilidad.

### Pregunta de Análisis 1 (Responde en una celda de texto):
Al analizar un algoritmo de ordenamiento sobre una lista, ¿qué escenarios representarían el "caso peor" y el "caso mejor" de los datos de entrada según la literatura revisada?

**RESPUESTA:** De acuerdo con la literatura de análisis de algoritmos (como Cormen et al. y Brassard & Bratley), al evaluar un algoritmo de ordenamiento basado en comparaciones (por ejemplo, *Insertion Sort* o *Bubble Sort*), estos escenarios dependen directamente del orden inicial de los datos de entrada:

* **Mejor caso (*Best Case*):** Ocurre cuando la lista de entrada **ya se encuentra completamente ordenada** según el criterio requerido (por ejemplo, de forma ascendente). El algoritmo solo realiza un recorrido de verificación haciendo $n - 1$ comparaciones y $0$ intercambios o desplazamientos, logrando una complejidad lineal óptima de $\Omega(n)$ (o $O(n)$).
* **Peor caso (*Worst Case*):** Ocurre cuando la lista de entrada está **ordenada en sentido completamente inverso** (descendente). Cada elemento debe compararse y desplazarse frente a todos los anteriores, ejecutando el máximo número posible de operaciones: $\sum_{i=1}^{n-1} i = \frac{n(n-1)}{2}$, lo que resulta en una complejidad cuadrática $O(n^2)$. *(Nota: En algoritmos como Quicksort con pivote fijo en un extremo, tanto una lista ya ordenada como una inversamente ordenada producen particiones totalmente desbalanceadas, degradándolo también a su peor caso $O(n^2)$).*

### Instrucciones:
1. Implementa una versión recursiva para el cálculo del factorial ($n! = n \cdot (n-1)!$) y una versión iterativa para Fibonacci.

````python
def factorial_recursivo(n):
    if n < 0:
        raise ValueError("El factorial no está definido para numeros negativos")
    if n == 0 or n == 1:
        return 1
    return n * factorial_recursivo(n - 1)

def fibonacci_iterativo(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    a, b = 0, 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b


print("Factorial recursivo (5!):", factorial_recursivo(5))
print("Fibonacci iterativo (n=10):", fibonacci_iterativo(10))
````

2. Utiliza la librería `time` o la función mágica `%timeit` para comparar el tiempo consumido (A posteriori) entre el Fibonacci recursivo y el iterativo para `n=35`.

````python
import time

n = 35


inicio_rec = time.time()
res_rec = fibonacci_recursivo(n)
fin_rec = time.time()
tiempo_rec = fin_rec - inicio_rec


inicio_ite = time.time()
res_ite = fibonacci_iterativo(n)
fin_ite = time.time()
tiempo_ite = fin_ite - inicio_ite

print(f"--- Comparación Empírica (Prueba A Posteriori) para n = {n} ---")
print(f"Fibonacci Recursivo -> Resultado: {res_rec} | Tiempo: {tiempo_rec:.6f} segundos")
print(f"Fibonacci Iterativo -> Resultado: {res_ite} | Tiempo: {tiempo_ite:.6f} segundos")
if tiempo_ite > 0:
    print(f"La version iterativa fue {tiempo_rec / tiempo_ite:,.0f} veces ms rápida")
````

| Criterio | Fibonacci Iterativo | Fibonacci Recursivo |
| :--- | :--- | :--- |
| **Complejidad Temporal** | $O(n)$ — Crecimiento lineal (polinomial de grado 1) | $O(2^n)$ o $O(\phi^n)$ — Crecimiento exponencial ($\phi \approx 1.618$) |
| **Complejidad Espacial** | $O(1)$ — Uso constante de memoria (solo 3 variables escalares) | $O(n)$ — Profundidad máxima de la pila de llamadas (*Call Stack*) |
| **Gestión de Subproblemas** | No repite cálculos; reutiliza los dos estados anteriores | Recalcula los mismos subproblemas de forma redundante |
| **Rendimiento empírico ($n=35$)** | Prácticamente instantáneo (microsegundos) | Tarda varios segundos y se vuelve intratable si crece $n$ |

**Conclusión:**
La versión **iterativa** crece de forma **polinomial** (específicamente lineal, $O(n)$) porque emplea un único ciclo `for` que avanza secuencialmente desde $2$ hasta $n$, ejecutando una cantidad constante $O(1)$ de operaciones por iteración sin repetir cálculos previos.
